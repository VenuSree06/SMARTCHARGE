```python
registrations = pd.read_csv(
    "datasets/raw/ev_cat_01-24.csv"
)

print("EV Registration Dataset Loaded!")
print("Shape:", registrations.shape)

display(registrations.head())

print("\nMissing Values:")
display(registrations.isnull().sum())

print("\nDuplicate Records:", registrations.duplicated().sum())
```

```python
print("First 10 Records:")
display(registrations.head(10))

print("\nLast 10 Records:")
display(registrations.tail(10))

print("\nData Types:")
print(registrations.dtypes)
```

```python
# Remove invalid first row
registrations = registrations[registrations['Date'] != '0'].copy()

# Convert Date column
registrations['Date'] = pd.to_datetime(
    registrations['Date'],
    format='%d/%m/%y',
    errors='coerce'
)

# Save cleaned dataset
registrations.to_csv(
    "datasets/processed/ev_registrations_cleaned.csv",
    index=False
)

print("EV Registration Dataset Cleaned Successfully!")
print("Shape:", registrations.shape)

display(registrations.head())
```

```python
# Calculate total registrations across vehicle categories

vehicle_columns = registrations.select_dtypes(include='number').columns

registrations['Total_Registrations'] = registrations[vehicle_columns].sum(axis=1)

# Extract year and month
registrations['Year'] = registrations['Date'].dt.year
registrations['Month'] = registrations['Date'].dt.month

print("Total EV Registrations:", registrations['Total_Registrations'].sum())

display(registrations[['Date', 'Year', 'Month', 'Total_Registrations']].head())
```

```python
yearly_ev = registrations.groupby('Year')['Total_Registrations'].sum().reset_index()

display(yearly_ev)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4))

plt.plot(
    yearly_ev['Year'],
    yearly_ev['Total_Registrations'],
    marker='o'
)

plt.title('Year-wise EV Registration Trend in India')
plt.xlabel('Year')
plt.ylabel('Total EV Registrations')
plt.grid(True)
plt.xticks(rotation=34)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

plt.plot(
    yearly_ev['Year'],
    yearly_ev['Total_Registrations'],
    color='#10B981',
    marker='o',
    markersize=6,
    linewidth=2.5
)

plt.title(
    "India's EV Registration Trend (2001–2024)",
    fontsize=17,
    fontweight='bold',
    color='#142D4E'
)

plt.xlabel("Year", fontsize=12)
plt.ylabel("Total EV Registrations", fontsize=12)

plt.xticks(yearly_ev['Year'], rotation=34)

plt.grid(
    True,
    linestyle='--',
    alpha=0.3
)

plt.ticklabel_format(
    style='plain',
    axis='y'
)

plt.tight_layout()
plt.show()
```

```python
yearly_ev['YoY_Growth_%'] = (
    yearly_ev['Total_Registrations'].pct_change() * 100
)

display(yearly_ev)
```

```python
# Import required libraries
import matplotlib.pyplot as plt
import numpy as np

# Remove the first year because its YoY growth is NaN
growth_data = yearly_ev.dropna(subset=['YoY_Growth_%'])

# Create a figure with suitable dimensions
plt.figure(figsize=(8, 6))

# Assign green to positive growth and red to negative growth
colors = [
    '#10B981' if value >= 0 else '#EF4444'
    for value in growth_data['YoY_Growth_%']
]

# Create the bar chart
plt.bar(
    growth_data['Year'],
    growth_data['YoY_Growth_%'],
    color=colors,
    width=0.7
)

# Add chart title
plt.title(
    'Year-over-Year EV Registration Growth in India',
    fontsize=17,
    fontweight='bold',
    color='#142D4E'
)

# Add axis labels
plt.xlabel('Year', fontsize=12)
plt.ylabel('YoY Growth (%)', fontsize=12)

# Add a horizontal zero line to separate growth and decline
plt.axhline(
    y=0,
    color='#263445',
    linewidth=1
)

# Display year labels
plt.xticks(growth_data['Year'], rotation=34)

# Add grid lines
plt.grid(
    axis='y',
    linestyle='--',
    alpha=0.3
)

# Adjust spacing
plt.tight_layout()

# Display the chart
plt.show()
```

```python
# Find the year with the highest YoY growth
highest_growth = growth_data.loc[
    growth_data['YoY_Growth_%'].idxmax()
]

# Find the year with the lowest YoY growth
lowest_growth = growth_data.loc[
    growth_data['YoY_Growth_%'].idxmin()
]

# Display the results
print("Highest EV Registration Growth:")
print(highest_growth)

print("\nLowest EV Registration Growth:")
print(lowest_growth)
```

```python
# Import pandas for data analysis
import pandas as pd

# Load the cleaned EV manufacturer dataset
makers_clean = pd.read_csv(
    'datasets/processed/ev_makers_cleaned.csv'
)

# Display the first five records
display(makers_clean.head())

# Check the dataset dimensions
print("Dataset Shape:", makers_clean.shape)

# Display column information
makers_clean.info()
```

```python
# Group manufacturers by state
# nunique() counts unique EV manufacturers in each state

state_manufacturers = (
    makers_clean.groupby('State')['EV Maker']
    .nunique()
    .reset_index()
)

# Rename columns for better understanding
state_manufacturers.columns = [
    'State',
    'Total_Manufacturers'
]

# Sort states from highest to lowest manufacturer count
state_manufacturers = state_manufacturers.sort_values(
    by='Total_Manufacturers',
    ascending=False
)

# Display the results
display(state_manufacturers)
```

```python
# Import matplotlib for data visualization
import matplotlib.pyplot as plt

# Create a figure with suitable size
plt.figure(figsize=(8, 5))

# Create a horizontal bar chart
bars = plt.barh(
    state_manufacturers['State'],
    state_manufacturers['Total_Manufacturers'],
    color='#10B981',
    edgecolor='#142D4E'
)

# Add the number of manufacturers at the end of each bar
for bar in bars:
    width = bar.get_width()
    plt.text(
        width + 0.2,
        bar.get_y() + bar.get_height() / 2,
        int(width),
        va='center',
        fontsize=10
    )

# Add chart title
plt.title(
    'EV Manufacturer Distribution by State',
    fontsize=16,
    fontweight='bold',
    color='#142D4E',
    pad=15
)

# Add axis labels
plt.xlabel('Number of EV Manufacturers', fontsize=12)
plt.ylabel('State', fontsize=12)

# Display states with the highest count at the top
plt.gca().invert_yaxis()

# Add grid lines for better readability
plt.grid(axis='x', linestyle='--', alpha=0.3)

# Remove unnecessary borders
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

# Adjust layout and display the chart
plt.tight_layout()
plt.show()
```
